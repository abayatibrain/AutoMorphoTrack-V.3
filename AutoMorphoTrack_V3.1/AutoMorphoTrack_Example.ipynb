# AutoMorphoTrack Example Notebook
This notebook demonstrates how to use the AutoMorphoTrack package, including detection, morphology classification, and lysosomal counting.

```python
# Install local package if needed
# !pip install -e ./AutoMorphoTrack
```

```python
import tifffile as tiff
import numpy as np
from AutoMorphoTrack.core import run_pipeline
```

```python
# Load your multi-frame, two-channel TIFF file
image_path = 'path/to/your_image.tif'
stack = tiff.imread(image_path)
# Optional: if the image is 4D (t, c, h, w), reshape to (t, h, w, c) or merge channels as needed
# Example: stack = np.moveaxis(stack, 1, -1) if needed
```

```python
# Define output directory and frame interval (in seconds, or arbitrary units)
output_dir = 'results/'
frame_interval = 1  # Adjust as needed
```

```python
# Run the AutoMorphoTrack pipeline
results = run_pipeline(stack, output_dir, frame_interval)
```

### Outputs
- Detection overlay: `detection_overlay.png`
- Mitochondrial morphology classification: `mitochondria_morphology.png`
- Lysosomal count and labeling: `lysosome_count_frame0.png`
- `results` dictionary contains raw outputs from morphology and lysosomal count steps.
